# Roadmap Portfolio Color Palette
Youssef Sherif — breedlove.xyz-style portfolio colors

```python
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import numpy as np

fig, ax = plt.subplots(figsize=(16, 6))
ax.set_xlim(0, 10)
ax.set_ylim(0, 4)
ax.axis('off')
fig.patch.set_facecolor('#0a0806')

colors = [
    ('Deep Background', '#0a0806'),
    ('Card Surface', '#111111'),
    ('Primary Text', '#f5f0e8'),
    ('Secondary Text', '#9a9590'),
    ('Gold Accent', '#c4a882'),
    ('Border', '#1a1816'),
]

for i, (name, hex_color) in enumerate(colors):
    x = 0.5 + i * 1.6
    rect = mpatches.FancyBboxPatch((x, 1.0), 1.3, 1.8, boxstyle='round,pad=0.05',
                                    facecolor=hex_color, edgecolor='#333333', linewidth=1)
    ax.add_patch(rect)
    ax.text(x + 0.65, 0.7, name, ha='center', va='top', fontsize=9, color='#888888',
            family='monospace')
    ax.text(x + 0.65, 0.35, hex_color, ha='center', va='top', fontsize=10, color=hex_color,
            family='monospace', weight='bold')

ax.set_title('Roadmap Portfolio — Color Palette', fontsize=18, color='#f5f0e8', pad=20)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import numpy as np

fig, axes = plt.subplots(1, 3, figsize=(16, 5))
fig.patch.set_facecolor('#0a0806')

data = [
    ('Text Hierarchy', ['#f5f0e8', '#c4a882', '#9a9590', '#555555']),
    ('Surface Colors', ['#0a0806', '#111111', '#1a1816', '#222222']),
    ('Accent & Border', ['#c4a882', '#c4a882', '#1a1816', '#333333']),
]

for ax, (title, pal) in zip(axes, data):
    ax.set_facecolor('#0a0806')
    ax.axis('off')
    ax.set_title(title, fontsize=12, color='#f5f0e8', pad=12)
    for i, c in enumerate(pal):
        rect = mpatches.FancyBboxPatch((0.1 + i * 0.24, 0.2), 0.22, 0.6,
                                        boxstyle='round,pad=0.02', facecolor=c,
                                        edgecolor='#333333', linewidth=1)
        ax.add_patch(rect)
        ax.text(0.1 + i * 0.24 + 0.11, 0.08, c, ha='center', va='top',
                fontsize=8, color=c, family='monospace')

plt.tight_layout()
plt.show()
```
